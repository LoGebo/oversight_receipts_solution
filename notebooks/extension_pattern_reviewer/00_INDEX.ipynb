## Result in one line (honest)
An LLM used as a second-line reviewer **does** catch pattern frauds that the engine cannot see (Sonnet 5.5 with 4 examples ≈ 88% recall with 0% false alarms on the independent test). The fine-tuned Qwen3-4B **won on our own benchmark** but **lost on the independent test** (it learned the generator's style): a negative result, with a diagnosis and a plan.
